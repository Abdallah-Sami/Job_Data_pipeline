```python
import pandas as pd
```

```python
df = pd.read_json("../data/raw/sabbar_2026-09-10.json")
df
```

```python
df.shape
```

```python
df.duplicated(subset=["url"]).sum()
```

```python
df = df.drop_duplicates(subset=["url"]).reset_index(drop=True)
```

```python
df.shape
```

```python
profile = pd.DataFrame({
    "dtype": df.dtypes,
    "nulls": df.isnull().sum(),
    "null_pct": (df.isnull().mean()*100).round(1),
    "unique": df.nunique(),
})
profile
```

```python
df_clean = df.copy()   
```

```python
df_clean["created_date"].dropna().head()
```

```python
df_clean["posted_date"] = pd.to_datetime(
    df_clean["created_date"], unit="s"
).dt.strftime("%Y-%m-%d")

df_clean[["created_date", "posted_date"]].head()
```

```python
df_clean.head(20)
```

```python
df_clean.duplicated().sum()
```

```python
df_clean = df_clean.rename(columns={
    "url": "job_url",
    "title": "job_title",
    "company": "company_name",
    "description": "job_description",
    "contract_type": "employment_type",
    "workplace_type": "work_mode",
    "experience_years": "experience_needed",
})

df_clean["source"] = "sabbar"
df_clean["city"] = df_clean["city"].replace(r"^\s*$", None, regex=True)

# دمج الراتب لنص (حسب الـSchema)
df_clean["salary"] = df_clean.apply(
    lambda r: f"{int(r['salary'])}-{int(r['max_salary'])}"
    if pd.notna(r["salary"]) and pd.notna(r["max_salary"])
    else (str(int(r["salary"])) if pd.notna(r["salary"]) else None),
    axis=1,
)


for col in ["career_level", "qualification_required", "skills",
            "estimated_salary", "collected_at"]:
    df_clean[col] = None


schema_cols = [
    "source", "job_title", "company_name", "job_description", "job_url",
    "country", "city", "work_mode", "employment_type", "career_level",
    "experience_needed", "qualification_required", "skills",
    "salary", "estimated_salary", "posted_date", "collected_at",
]
df_clean = df_clean[schema_cols]
df_clean.head()
```

```python
df_clean
```

```python
df_clean.to_csv("../data/interim/cleaned_sabbar.csv", index=False, encoding="utf-8-sig")
print("saved:", df_clean.shape)
```

```python
df_clean.to_json(
    "../data/interim/cleaned_sabbar.json",
    orient="records",
    force_ascii=False,
    indent=2,
)
print("saved json:", df_clean.shape)
```
